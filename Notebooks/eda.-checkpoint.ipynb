```python
def get_directory_tree(root, max_depth=3):
    root = Path(root)

    for path in sorted(root.rglob("*")):
        relative = path.relative_to(root)
        depth = len(relative.parts) - 1

        if depth <= max_depth:
            prefix = "    " * depth

            if path.is_dir():
                print(f"{prefix}[DIR]  {path.name}")
            else:
                print(f"{prefix}[FILE] {path.name}")


get_directory_tree(DATASET_DIR, max_depth=3)
```

```python
def infer_class_from_path(path):
    """
    Infer dataset class from directory/file naming.
    Adjust this mapping after inspecting the actual dataset tree.
    """

    text = str(path).lower()

    if "covid" in text:
        return "COVID-19"

    if "non-covid" in text or "non_covid" in text:
        return "Non-COVID Infection"

    if "normal" in text:
        return "Normal"

    return "Unknown"
```

```python
records = []

for image_path in tqdm(
    xray_candidates,
    desc="Building image metadata"
):

    try:
        with Image.open(image_path) as img:

            width, height = img.size
            mode = img.mode
            format_name = img.format
            channels = len(img.getbands())

        file_size = image_path.stat().st_size

        records.append({
            "path": str(image_path),
            "filename": image_path.name,
            "class": infer_class_from_path(image_path),
            "width": width,
            "height": height,
            "channels": channels,
            "mode": mode,
            "format": format_name,
            "file_size_bytes": file_size,
            "file_size_mb": file_size / (1024 ** 2)
        })

    except Exception as e:

        records.append({
            "path": str(image_path),
            "filename": image_path.name,
            "class": infer_class_from_path(image_path),
            "width": None,
            "height": None,
            "channels": None,
            "mode": None,
            "format": None,
            "file_size_bytes": None,
            "file_size_mb": None
        })

metadata_df = pd.DataFrame(records)
```

```python
def show_class_montage(
    df,
    class_name,
    n=12,
    figsize=(16, 12)
):

    subset = df[
        df["class"] == class_name
    ]

    sample = subset.sample(
        min(n, len(subset)),
        random_state=SEED
    )

    cols = 4
    rows = math.ceil(len(sample) / cols)

    fig, axes = plt.subplots(
        rows,
        cols,
        figsize=figsize
    )

    axes = np.array(axes).reshape(-1)

    for ax, (_, row) in zip(
        axes,
        sample.iterrows()
    ):

        try:
            img = Image.open(row["path"]).convert("L")

            ax.imshow(img, cmap="gray")
            ax.set_title(
                f"{class_name}\n"
                f"{row['width']}×{row['height']}"
            )
            ax.axis("off")

        except Exception:
            ax.axis("off")

    for ax in axes[len(sample):]:
        ax.axis("off")

    plt.suptitle(
        f"{class_name} — Sample Images",
        fontsize=16
    )

    plt.tight_layout()
    plt.show()
```

```python
def find_matching_mask(image_path, mask_files):
    """
    Attempts to identify the corresponding mask
    based on filename similarity.
    """

    image_stem = Path(image_path).stem.lower()

    candidates = []

    for mask_path in mask_files:

        mask_stem = Path(mask_path).stem.lower()

        if image_stem in mask_stem or mask_stem in image_stem:
            candidates.append(mask_path)

    return candidates[0] if candidates else None
```

```python
def show_xray_mask_overlay(
    image_path,
    mask_path
):

    image = np.asarray(
        Image.open(image_path).convert("L")
    )

    mask = np.asarray(
        Image.open(mask_path).convert("L")
    )

    plt.figure(figsize=(8, 8))

    plt.imshow(
        image,
        cmap="gray"
    )

    plt.imshow(
        mask,
        cmap="jet",
        alpha=0.35
    )

    plt.title(
        f"X-Ray + Lung Mask\n{Path(image_path).name}"
    )

    plt.axis("off")
    plt.tight_layout()
    plt.show()
```

```python
def calculate_sha256(file_path, chunk_size=1024 * 1024):

    sha256 = hashlib.sha256()

    with open(file_path, "rb") as f:

        while chunk := f.read(chunk_size):
            sha256.update(chunk)

    return sha256.hexdigest()
```

```python
hash_records = []

for image_path in tqdm(
    xray_candidates,
    desc="Hashing images"
):

    try:

        file_hash = calculate_sha256(
            image_path
        )

        hash_records.append({
            "path": str(image_path),
            "sha256": file_hash
        })

    except Exception:
        pass

hash_df = pd.DataFrame(hash_records)
```
