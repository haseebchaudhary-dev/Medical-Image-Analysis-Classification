```python
def normalize_stem(path: Path):

    stem = path.stem.lower()

    replacements = [
        "_mask",
        "-mask",
        " mask",
        "_lung",
        "-lung",
        " lung",
        "_segmentation",
        "-segmentation",
        " segmentation"
    ]

    for replacement in replacements:
        stem = stem.replace(replacement, "")

    stem = re.sub(
        r"[^a-z0-9]+",
        "_",
        stem
    )

    stem = stem.strip("_")

    return stem
```

```python


from pathlib import Path
import hashlib
import pandas as pd
from tqdm.auto import tqdm


def sha256_file(path, chunk_size=1024 * 1024):
    """
    Calculate SHA-256 hash for a file.

    Parameters
    ----------
    path : str or pathlib.Path
        File path.
    chunk_size : int
        Number of bytes read per iteration.

    Returns
    -------
    str
        SHA-256 hexadecimal hash.
    """

    path = Path(path)

    sha256 = hashlib.sha256()

    with path.open("rb") as f:
        while True:
            chunk = f.read(chunk_size)

            if not chunk:
                break

            sha256.update(chunk)

    return sha256.hexdigest()




hash_records = []
hash_errors = []

for path in tqdm(xray_files, desc="Hashing X-rays"):

    path = Path(path)

    try:

        file_hash = sha256_file(path)

        hash_records.append({
            "path": str(path),
            "sha256": file_hash
        })

    except Exception as e:

        hash_errors.append({
            "path": str(path),
            "error": str(e)
        })

        print(f"Hash error: {path} -> {e}")



hash_df = pd.DataFrame(
    hash_records,
    columns=["path", "sha256"]
)

hash_error_df = pd.DataFrame(
    hash_errors,
    columns=["path", "error"]
)




total_files = len(xray_files)
successful_files = len(hash_df)
failed_files = len(hash_error_df)

print("\n" + "=" * 60)
print("SHA-256 HASHING SUMMARY")
print("=" * 60)

print(f"Total X-ray files : {total_files:,}")
print(f"Successfully hashed: {successful_files:,}")
print(f"Hashing failures   : {failed_files:,}")




if not hash_df.empty:

    duplicate_hash_df = (
        hash_df[
            hash_df["sha256"].duplicated(keep=False)
        ]
        .sort_values("sha256")
        .reset_index(drop=True)
    )

    duplicate_hash_groups = (
        hash_df["sha256"]
        .value_counts()
        .loc[lambda x: x > 1]
    )

else:

    duplicate_hash_df = pd.DataFrame(
        columns=["path", "sha256"]
    )

    duplicate_hash_groups = pd.Series(
        dtype="int64"
    )


print(f"Exact duplicate files : {len(duplicate_hash_df):,}")
print(f"Duplicate hash groups : {len(duplicate_hash_groups):,}")




print("\nHash dataframe:")
display(hash_df.head())

print("\nExact duplicates:")
display(duplicate_hash_df.head(20))
```

```python
SOURCE_KEYWORDS = {
    "covid_qu": [
        "covid_qu",
        "covidqu"
    ],

    "qata_cov19": [
        "qata",
        "qatacov"
    ],

    "rsna": [
        "rsna"
    ],

    "padchest": [
        "padchest"
    ],

    "pneumonia_xray": [
        "pneumonia"
    ],

    "cxnet": [
        "cxnet"
    ]
}
```

```python
def infer_source(path):

    text = str(path).lower()

    matches = []

    for source, keywords in SOURCE_KEYWORDS.items():

        if any(
            keyword in text
            for keyword in keywords
        ):
            matches.append(source)

    if len(matches) == 1:
        return matches[0]

    if len(matches) > 1:
        return "multiple_possible"

    return "unknown"
```

```python
def extract_possible_patient_id(filename):

    stem = Path(filename).stem.lower()

    patterns = [
        r"patient[_\-]?([a-z0-9]+)",
        r"patient[_\-]?id[_\-]?([a-z0-9]+)",
        r"case[_\-]?([a-z0-9]+)",
        r"subject[_\-]?([a-z0-9]+)",
        r"person[_\-]?([a-z0-9]+)"
    ]

    for pattern in patterns:

        match = re.search(
            pattern,
            stem
        )

        if match:
            return match.group(1)

    return None
```

```python
xray_inventory["possible_patient_id"] = (
    xray_inventory["filename"]
    .apply(extract_possible_patient_id)
)
```

```python
def build_group_id(row):

    patient_id = row[
        "possible_patient_id"
    ]

    if pd.notna(patient_id):
        return f"patient_{patient_id}"

    source = row[
        "source_guess"
    ]

    sample_id = row[
        "sample_id"
    ]

    return f"{source}_{sample_id}"
```

```python
xray_inventory["group_id"] = (
    xray_inventory
    .apply(build_group_id, axis=1)
)
```
